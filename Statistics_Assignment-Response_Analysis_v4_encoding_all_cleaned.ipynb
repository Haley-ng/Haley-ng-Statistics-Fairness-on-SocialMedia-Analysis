## Rank personal values
Huyen:

turn the survey question "What matters most to you?" in into four number columns:

- rank_autonomy
- rank_democracy
- rank_equality
- rank_equity

each answer is a text listing the 4 options in the order the person chose them, separated by ';'.
The position of an option in that list is its rank (1 = most important, 4 = least important).

#### 1: Look at the ranking question

```python
per_value = "What matters most to you? (1 being the most important and the last number being the least important)"

data[per_value].head()
```

look at just **one** answer to understand it:

```python
first_answer = data[per_value][0]
print(first_answer)
```

#### 2: Split values
`split(";")` cuts the text every time it sees a `;`.

```python
value_split = first_answer.split(";")
print(value_split)
```

```python
# The last piece is an empty text `''`, because the answer ends with `;`.
# I remove empty value with a simple loop

clean_value = []
for piece in value_split:
    piece = piece.strip()      # remove spaces at the start and end
    if piece != "":            # keep only pieces that are not empty
        clean_value.append(piece)

print(clean_value)
```

#### 3: Find the position of each value to rank

I go through the list and count the position, starting at 1.
I use `startswith` to check which option a value is.


```python
# Use elif to make each value match only one category. 
# Check conditions and stop when it find the first value matching condition. 
# Then increases the ranking position by 1 after each iteration.
# loop each row. if value starts with "Autonomy", then print position = 1 and next postion is position + 1, meaning new postions 2,3,4 
position = 1
for piece in clean_value:
    if piece.startswith("Autonomy"):            
        print("Autonomy has rank", position)
    elif piece.startswith("Democratic"):
        print("Democracy has rank", position)
    elif piece.startswith("Equality"):
        print("Equality has rank", position)
    elif piece.startswith("Equity"):
        print("Equity has rank", position)
    position = position + 1
```

#### 4: Do the same for every respondent

Now I repeat this for all rows. I make 4 empty lists, one per new column.
For each answer I add one number to each list.

Some people did not answer the question. Their cell is empty (`NaN`), so I add `None` (missing) to all 4 lists.

```python
autonomy_list = []
democracy_list = []
equality_list = []
equity_list = []

for answer in data[per_value]:

    # Case 1: the person did not answer
    if pd.isna(answer):
        autonomy_list.append(None)
        democracy_list.append(None)
        equality_list.append(None)
        equity_list.append(None)
        continue    # go to the next answer

    # Case 2: the person answered, then split and clean
    clean_pieces = []
    for piece in answer.split(";"):
        piece = piece.strip()
        if piece != "":
            clean_pieces.append(piece)

    # Find the rank of each option
    rank_autonomy = None
    rank_democracy = None
    rank_equality = None
    rank_equity = None

    position = 1
    for piece in clean_pieces:
        if piece.startswith("Autonomy"):
            rank_autonomy = position
        elif piece.startswith("Democratic"):
            rank_democracy = position
        elif piece.startswith("Equality"):
            rank_equality = position
        elif piece.startswith("Equity"):
            rank_equity = position
        position = position + 1

    # Save the ranks of this person
    autonomy_list.append(rank_autonomy)
    democracy_list.append(rank_democracy)
    equality_list.append(rank_equality)
    equity_list.append(rank_equity)

print("Done! Number of answers:", len(autonomy_list))
```

#### 5: Put the lists into a new dataframe

```python
ranks = pd.DataFrame()
ranks["ID"] = data["ID"]
ranks["rank_autonomy"] = autonomy_list
ranks["rank_democracy"] = democracy_list
ranks["rank_equality"] = equality_list
ranks["rank_equity"] = equity_list

ranks.head(10)
```

# ENCODING (one-hot & ordinal)
#Author Daniil Malygin

Statistical models only work with numbers, so every text answer has to be encoded. How we encode depends on the type of variable:

| Type | Meaning | Variables | Encoding |
|---|---|---|---|
| **Nominal** | categories without an order | country, parents' origin, gender, online languages, top-ranked value, wave | **One-hot**: one 0/1 column per category |
| **Ordinal** | categories with a natural order | Likert answers, usage frequency, education, age | **Ordinal codes** that keep the order (e.g. 1–5) |

Three rules we apply:
1. **Missing stays missing.** A respondent without an answer gets `NaN` in every dummy, not 0. Otherwise they would silently be counted as the reference category.
2. **Dummy-variable trap.** The dummies of one variable always add up to 1, which is perfectly collinear with the regression intercept. In a model we therefore **leave one dummy out** (the *reference category*) and interpret the others relative to it. We keep all dummies in the data and choose the reference at model time (`REFERENCE` at the end of this section).
3. **Fixed category lists**, so the same columns always exist in the same order.

```python
#This is a helpeer function that helps to one-hot encode one column with a fixed list of categories.

# pd.Categorical fixes the categories (and their order); get_dummies makes one 0/1 column per category.
# Rows where the original value is missing get <NA> in every dummy instead of 0
def one_hot(series, categories, prefix):
    dummies = pd.get_dummies(pd.Categorical(series, categories=categories), prefix=prefix, dtype=int)
    dummies.index = series.index                                   # keep the respondents' row index
    dummies = dummies.astype("Int64")                              # integer type that allows <NA>
    dummies.loc[series.isna(), :] = pd.NA                          # missing answer -> missing dummies
    return dummies
```

## 1. Country of residence → one-hot
The study is about four countries, so we keep those and group every other country into **Other** (44 respondents; most analyses will exclude them, but the dummy makes that choice explicit).

```python
# Daniil:
STUDY_COUNTRIES = ["Netherlands", "Germany", "Spain", "Greece"]

# countries outside the four study countries -> "Other"; missing stays missing
data["country_group"] = data["country_located"].where(data["country_located"].isin(STUDY_COUNTRIES) | data["country_located"].isna(), "Other")

country_dummies = one_hot(data["country_group"], STUDY_COUNTRIES + ["Other"], prefix="country_located")
data = pd.concat([data, country_dummies], axis=1)

print(data["country_group"].value_counts(dropna=False))
country_dummies.head()
```

## 2. Parents' origin → 4 groups → one-hot
After the spelling clean-up there are still 9 answer texts, because each country's form worded the options differently. They describe **4 situations**, the same labels as in `Responses_cleaned.xlsx`:
- **Native**: both parents born and brought up in the country of the form (NL, DE, ES or GR)
- **Both other-EU**: both parents from another EU country
- **Both non-EU**: both parents from outside the EU
- **Mixed EU/non-EU**: one parent from the EU, one from outside

The paper compares only **2 groups**: both parents from the EU (Native + other-EU) vs. at least one parent from outside the EU. A two-category variable needs just **one** dummy: `parents_non_eu` (1 = at least one parent from outside the EU).

```python
# Daniil:
# parent_origin was already standardised into the 4 groups in the cleaning step (Floris, cell above),
# so no second mapping is needed - parent_group simply takes those values.
PARENT_ORDER = ["Native", "Both other-EU", "Both non-EU", "Mixed EU/non-EU"]
data["parent_group"] = data["parent_origin"]

# check: every non-missing answer must already be one of the 4 groups
unmapped = data.loc[data["parent_group"].notna() & ~data["parent_group"].isin(PARENT_ORDER), "parent_group"].unique()
assert len(unmapped) == 0, f"unexpected values in parent_origin: {unmapped}"

parent_dummies = one_hot(data["parent_group"], PARENT_ORDER, prefix="parents")
data = pd.concat([data, parent_dummies], axis=1)

# paper's 2-group version: one dummy is enough for a binary variable
data["parents_non_eu"] = data["parent_group"].isin(["Both non-EU", "Mixed EU/non-EU"]).astype("Int64")
data.loc[data["parent_group"].isna(), "parents_non_eu"] = pd.NA

print(pd.crosstab(data["parent_origin"], data["parent_group"]))   # shows exactly which answer went where
print("\nAt least one non-EU parent:\n", data["parents_non_eu"].value_counts(dropna=False))
```

## 4. Online languages → multi-hot
This question allowed **several answers** (`"English;Spanish;"`), so one respondent can have several 1s. This is *multi-hot* encoding: each language is its own yes/no variable. Because the columns do not add up to 1, there is no dummy-variable trap here.

The free-text "other" answers need a spelling fix first ("Germany", "Deutsch" → German; "Greece" → Greek; "Català", "Catalán" → Catalan). Languages used by fewer than 15 respondents are grouped as **other**.

```python

# Huyen:
# If people use English on social media, then assign 1, else 0 (missing answer stays missing)
# True whenever "English" appears anywhere in the answer
# new column, so the original text in "online_english" is kept for the language dummies below
data["uses_english_online"] = (
    data["online_english"].str.contains("english", case=False, na=False).astype("Int64")
)
data.loc[data["online_english"].isna(), "uses_english_online"] = pd.NA

print(data["uses_english_online"].value_counts(dropna=False))
```

```python
# Daniil:
LANGUAGE_FIX = {"germany": "german", "deutsch": "german", "greece": "greek", "netherlands": "dutch",
                "català": "catalan", "catalán": "catalan", "portugese": "portuguese", "التركية": "turkish"}

# one row per (respondent, language): split on ';', clean, drop the empty piece after the last ';'
languages = data["online_english"].str.split(";").explode().str.strip().str.lower()
languages = languages[languages.notna() & (languages != "")].replace(LANGUAGE_FIX)

# keep languages with at least 15 users, the rest -> "other"
counts = languages.value_counts()
main_languages = counts[counts >= 15].index
languages = languages.where(languages.isin(main_languages), "other")
print(languages.value_counts())

# crosstab gives respondents x languages counts; clip at 1 -> 0/1
lang_dummies = pd.crosstab(languages.index, languages).clip(upper=1).add_prefix("lang_").rename_axis(index=None, columns=None)
lang_dummies = lang_dummies.reindex(data.index).astype("Int64")      # respondents without an answer -> <NA>
data = pd.concat([data, lang_dummies], axis=1)
data["n_online_languages"] = lang_dummies.sum(axis=1, min_count=1)  # how many languages someone uses online

lang_dummies.head()
```

## 7. Ordinal encoding of the survey answers (not one-hot)
The 40 survey questions use four answer scales. Each gets a dictionary that keeps the order: higher number = more agreement / more often / longer.

| Scale | Coding | Items |
|---|---|---|
| Agreement | 1 Strongly disagree … 5 Strongly agree | most items |
| Frequency | 1 Never … 5 Always ("Very often" for DVP2) | USE, UCD2, UAE3, REVP1, REVP2, RAD4, RAD5, DVP2 |
| Duration | 1 Less than a day … 5 More than a week; "I don't know" → missing | UAE4 |
| Shadowban exposure | 1 No, never · 2 heard of it · 3 a few times · 4 many times | DVP5 |

Some answers contain a hidden non-breaking space (`"3-5 days\xa0"`), so we clean the text before mapping; otherwise those answers would silently become missing.

```python
# Daniil:
AGREE    = {"Strongly disagree": 1, "Disagree": 2, "Neutral": 3, "Agree": 4, "Strongly agree": 5}
FREQ     = {"Never": 1, "Rarely": 2, "Sometimes": 3, "Often": 4, "Always": 5, "Very often": 5}
DURATION = {"Less than a day": 1, "1-2 days": 2, "3-5 days": 3, "About a week": 4, "More than a week": 5,
            "I don't know": np.nan}                       # "don't know" is not a point on the scale
SHADOW   = {"No, never": 1, "I've heard of it but haven't seen it": 2,
            "Yes, a few times": 3, "Yes, many times": 4}

FREQ_ITEMS = ["USE", "UCD2", "UAE3", "REVP1", "REVP2", "RAD4", "RAD5", "DVP2"]
LIKERT_ITEMS = ["FAIR0", "UCD1", "UCD2", "UCD3", "UAE1", "UAE2", "UAE3", "UAE4",
                "REVP1", "REVP2", "REVP3", "REVP4", "REVP5", "REVP6", "BC1", "BC2", "BC3", "BC4", "BC5",
                "NCR1", "NCR2", "NCR3", "NCR4", "DVP1", "DVP2", "DVP3", "DVP4", "DVP5",
                "RAD1", "RAD2", "RAD3", "RAD4", "RAD5", "RAD6", "RAD7", "ON1", "ON2", "ON3", "PF1", "PF2"]

def scale_of(item):
    if item == "UAE4":
        return DURATION
    if item == "DVP5":
        return SHADOW
    if item in FREQ_ITEMS:
        return FREQ
    return AGREE

data["UAE4_dont_know"] = (data["UAE4"].str.strip() == "I don't know").astype(int)   # keep this information

lost = {}
for item in ["USE"] + LIKERT_ITEMS:
    text = data[item].str.replace("\xa0", " ").str.strip()         # remove hidden spaces
    data[item] = text.map(scale_of(item))
    became_missing = text.notna() & data[item].isna()
    if became_missing.any():
        lost[item] = text[became_missing].value_counts().to_dict()

print("Answers that did not become a number:", lost)              # expected: only UAE4 "I don't know"
data[["USE", "UCD1", "UCD2", "UAE4", "DVP5", "PF1"]].describe().loc[["count", "min", "max"]]
```

**Usage intensity.** 87% use social media "Often" or "Always", so the 1–5 scale has little spread. For group comparisons (SQ4) we split it into a binary dummy: **heavy user** = "Always" (about half the sample).

```python
# Daniil:
data["heavy_user"] = (data["USE"] == 5).astype("Int64")
data.loc[data["USE"].isna(), "heavy_user"] = pd.NA
data["heavy_user"].value_counts(dropna=False)
```

## 8. Education and age
**Education** is ordered, so it gets ordinal codes (1 = secondary or less … 4 = master or higher).

**Age** is ordered too, but the survey's age groups are not equally spaced (there is no 55–64 option) and most respondents are 18–34. We therefore collapse it into 4 groups and **one-hot** them, so a model doesn't assume each step up in age has the same effect.

```python
# Daniil:
EDUCATION_LEVEL = {"Secondary or less": 1, "Volcational": 2, "Vocational": 2, "Bachelor": 3, "Master +": 4}
data["education_level"] = data["education"].map(EDUCATION_LEVEL)   # accepts both spellings of "Vocational"

AGE_GROUP = {"Under 18": "Under 18", "18-24": "18-24", "25-34": "25-34",
             "35-44": "35+", "45-54": "35+", "65 and older": "35+"}
data["age_group"] = data["age"].map(AGE_GROUP)
age_dummies = one_hot(data["age_group"], ["Under 18", "18-24", "25-34", "35+"], prefix="age")
data = pd.concat([data, age_dummies], axis=1)

print(data["education_level"].value_counts(dropna=False).sort_index())
print(data["age_group"].value_counts(dropna=False))
```

## 9. Checks and reference categories
**Check 1:** within each one-hot block, every respondent with an answer must have exactly one 1. That this sum is always 1 is exactly why one dummy must be left out of a regression (dummy-variable trap).

**Reference categories** (left out in models; other dummies are interpreted relative to them):

| Variable | Reference | Why |
|---|---|---|
| Country | Netherlands | largest group; Mona's original sample |
| Parents' origin (4 groups) | Native | majority group |
| Gender | Male | the paper's most "advantaged" comparison group |
| Age | 18-24 | largest group |
| Top value | autonomy | most often ranked first |

```python
# Daniil:
blocks = {"country": country_dummies.columns, "parents": parent_dummies.columns, "gender": gender_dummies.columns,
          "top": top_dummies.columns, "age": age_dummies.columns}
for name, cols in blocks.items():
    row_sums = data[cols].sum(axis=1, min_count=1).dropna()
    print(f"{name:8s} dummies per answered respondent sum to 1: {(row_sums == 1).all()}  (n = {len(row_sums)})")
```

```python
# Daniil:
# Columns to use as predictors in a regression: all dummies except the reference one
REFERENCE = {"country": "country_Netherlands", "parents": "parents_Native", "gender": "gender_Male",
             "age": "age_18-24", "top": "top_autonomy"}

def model_dummies(name):
    return [c for c in blocks[name] if c != REFERENCE[name]]

print("Country dummies for a model:", model_dummies("country"))
print("Parent dummies for a model: ", model_dummies("parents"))

encoded = (list(country_dummies.columns) + list(parent_dummies.columns) + ["parents_non_eu"]
           + list(gender_dummies.columns) + list(lang_dummies.columns) + ["n_online_languages"]
           + list(top_dummies.columns) + ["wave2", "heavy_user", "education_level"] + list(age_dummies.columns))
print("\nNew encoded columns:", len(encoded))
data[encoded].head()
```

### Encoding - Conclusion
- Nominal variables are **one-hot encoded**: country (5), parents' origin (4, plus the paper's binary `parents_non_eu`), gender (3), top-ranked value (4) and age group (4). Online languages are **multi-hot** (several languages per person), and wave is a binary dummy.
- Ordered variables keep their order as **ordinal codes**: all 40 survey items (1–5; DVP5 1–4), usage frequency and education. `heavy_user` splits usage into "Always" vs. less.
- Missing answers stay missing in every dummy, and each model leaves out the reference category in `REFERENCE` to avoid the dummy-variable trap.
- Who uses what: country, `parents_*` and `wave2` → country/origin moderation (SQ4a); `gender_*`, `heavy_user`, `age_*` → user-profile moderation (SQ4b); survey items → scales for SQ1–3.
